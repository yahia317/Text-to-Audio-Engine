```python
def load_csv_files_nparray(cssv_file_path):
    # Load data from CSV file into a NumPy array
    data = np.genfromtxt(cssv_file_path, delimiter=',')

    # Print the array to verify the data has been loaded correctly
    return data
```

```python
def load_csv_files_dict(cssv_file_path):
    data_list = []
    csv.field_size_limit(1000000000)
    with open(cssv_file_path, 'r') as file:
        reader = csv.reader(file)
        for row in reader:
            data_list.append(row)
    return data_list
```

```python
def list_to_array_with_remove_of_first(my_list):
    del my_list[0]
    my_array = np.array(my_list)
    return my_array
```
