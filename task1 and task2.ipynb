```python
def get_fileName(captions_path):
    with open(captions_path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for row in csvreader:
            column_indices = 0  # Choose columns 1 and 4 (indices start from 0)
            selected_column = [row[column_indices] for row in csvreader]
    return selected_column
```

```python
def load_the_data(dataset_path, csv_path, captions_path):#, n_fft=1024, sr=44100, n_mels=16
    # dictionary to store mel_spec
    data = []
    max_len=0
    all_max = 264600
    
    all_max_captions_size = all_max/5 #all caption should have this size
    file_names = get_fileName(captions_path)
    # loop through all sub-folders
    number_of_audio_files = len(file_names)
    for i in range(number_of_audio_files):
        # load audio files with librosa
        file_path = os.path.join(dataset_path, file_names[i])
        scale, sr = librosa.load(file_path, sr=None)
        #calculate samples_per_segment 
        samples_per_segment = len(scale)/5
        # process all 5 captions of audio file
        for d in range(5):
            # calculate start and finish sample for current segment
            start = math.floor(samples_per_segment * d) # with round 2.5 become 2 and 2.6 become 3
            finish = math.floor(start + samples_per_segment)
            a = scale[start:finish]
            padded_row = np.pad(a, ((0, all_max - a.shape[0])), mode='constant')
            log_mel_spect = librosa.power_to_db(librosa.feature.melspectrogram(y=padded_row, sr=sr, n_fft=1024, hop_length=512, n_mels=64))#librosa.power_to_db()
            maximum = 32.8195
            minimum = -100.0
            max_val = 1
            min_val = 0
            normalized_data = (log_mel_spect - minimum) / (maximum - minimum) * (max_val-min_val) + min_val
            data.append(normalized_data)
            #bach nsib lmax valeur
            leen = samples_per_segment
            if (leen>max_len):
                max_len=leen
                
    print("max : " , max_len)
    #cant convert data to np untill all element have same leng or we loss float32 to object
    print("data loaded succesfully")
    """
    #max and min for log mel spect
    maximum = 32.8195
    minimum = -100.0
    """
    """
    #max and min for mel spect 
    maximum = 1124.6125
    minimum = 0.0
    """
    data_array = np.array(data)
    data_array.shape
    """
    #nahi had 3 point e brk tjik normalizi
    data_norm = apply_normalization(np.array(data), maximum, minimum)
    del data
    """
    del data
    save_data_csv(data_array.reshape((-1, data_array.shape[-1])),csv_path)
```

```python
def max_leng_of_data(data_array):
    number_of_items = len(data_array)
    max_len=0
    for i in range(number_of_items):
        if (len(data_array[i])>max_len):
            max_len=len(data_array[i])
    print(max_len)
    return max_len
```

```python
def right_padding(array, max_nb_item):
    data = []
    number_elements = len(array)
    for i in range(number_elements):
        padded_array = np.pad(array[i], ((0, max_nb_item - array[i].shape[0])), mode='constant')#right padding
        #padded_array = np.pad(array[i], ((max_nb_item - array[i].shape[0], 0)), mode='constant')#left padding
        data.append(padded_array)
    print("data padded succesful")
    return data
    
```

```python
def apply_log_mel_spectrogram(padded_data, sr=44100, n_fft=1024, hop_length=512, n_mels=64):
    data = []
    data_len = len(padded_data)
    for i in range(data_len): 
        #y = padded_data[i] instead of just padded_data[i] wil help in some error
        log_mel_spect = librosa.feature.melspectrogram(y=padded_data[i], sr=sr, n_fft=n_fft, hop_length=hop_length, n_mels=n_mels)#librosa.power_to_db()
        data.append(log_mel_spect)#.tolist()
    print("log_mel_spec succesful")
    return data
```

```python
def apply_normalization(array, maximum, minimum, max_val=1, min_val=0):
    print("np.max(array) : " , np.max(array))
    print("np.min(array) : " , np.min(array))
    print("maximum : " , maximum)
    print("minimum : " , minimum)
    # Normalize the data between 0 and 1
    normalized_data = (array - minimum) / (maximum - minimum)
    normalized_data = normalized_data * (max_val-min_val) + min_val
    print("normalized_data succesful")
    return normalized_data.reshape((-1, normalized_data.shape[-1]))
    
```

```python
def apply_denormalization(normalized_array, maximum, minimum, max_val=1, min_val=0):
    print("maximum : " , maximum)
    print("minimum : " , minimum)
    # Normalize the data between 0 and 1
    denormalized_data = (normalized_array - min_val) / (max_val - min_val)
    denormalized_data = denormalized_data * (maximum-minimum) + minimum
    return denormalized_data
```

```python
def read_all_captions(caption_path):
    with open(caption_path, 'r') as csvfile:
        csvreader = csv.reader(csvfile)
        for row in csvreader:
            column_index = 1  # index of the second column (0-based indexing)
            fifth_column = [row[column_index:] for row in csvreader]
    return fifth_column
```

```python
def model_Sbert(array_captions):
    model_name = "all-MPnet-base-v2"
    model = SentenceTransformer(model_name)
    embeddings = model.encode(array_captions, batch_size = 256, convert_to_numpy = True, show_progress_bar = True) 
    return embeddings
```

```python
def save_csv_captions(CSV_PATH_captions, embeddings_captions):
    with open(CSV_PATH_captions, "w") as csv_file:
        np.savetxt(CSV_PATH_captions, embeddings_captions, delimiter=',')
```

```python
def normalization_of_data(data_embeddings_captions):
    # Define the minimum and maximum values for scaling
    min_val = np.min(data_embeddings_captions)
    max_val = np.max(data_embeddings_captions)
    print(min_val)
    print(max_val)
    # Normalize the data between 0 and 1
    normalized_embeddings_captions = (data_embeddings_captions - min_val) / (max_val - min_val)
    return normalized_embeddings_captions
```

```python
def denormalization_of_data(data_embeddings_captions, normalized_embeddings_captions):
    
    # Define the minimum and maximum values of the original data
    min_val = np.min(data_embeddings_captions)
    max_val = np.max(data_embeddings_captions)

    # Denormalize the data
    denormalized_embeddings_captions = (normalized_embeddings_captions * (max_val - min_val)) + min_val
    return denormalized_embeddings_captions
```

```python
def checking_if_two_arrays_equal(embeddings_captions, denormalized_embeddings_captions):
    tolerance = 1e-6

    if np.all(np.abs(denormalized_embeddings_captions - embeddings_captions) < tolerance):
        print("The arrays are approximately equal.")
    else:
        print("The arrays are not equal.")
```
